# Reorganizing Array

changing the shape of matrix

```python
before=np.array([[1,2,3,4],[5,6,7,8]])
print(before)
after=np.reshape(before,(4,2))
print(after)
```

Any Matrix Converting into 1D array

```python
# ravel changes are made in main array
before=before.ravel()
before[0]=100
print(before)
print(after)
```

```python
# flatten changes are not made in main array
before=after.flatten()
before[0]=200
print(before)
print(after) 
```
