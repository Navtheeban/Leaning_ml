# Scatter Plot

```python
# The ways of styling themes as (set_style()) are as follows:
# white
# dark
# whitegrid
# darkgrid
# ticks

# set_context() allows us to override default parameters. This affects things like the size of the labels, 1nes, and other elements of the plot, but not the overall style -> set_style(). The contex are:
# poster 
# Paper Ideal for printing in academic papers.
# Notebook Good for working inside Jupyter notebooks (default setting).
# Talk Good for presentation slides where readability matters.

# palettte
# Set1'
# 'Set2'
# Set3
# "deep"
# "muted""
# "bright"
# "pastel"
# "dark"
# "colorblind"
# Qualitative
# Sequential
# Diverging
```

```python
sns.set_style('ticks')
sns.set_context('poster')
# data means dataframe name
# x means x-axis
# y means y-axis
# hue means identify group like which penguins are group in island
sns.scatterplot(data=penguins,x='flipper_length_mm',y='body_mass_g',hue='island',palette='Dark2')
# despine means the line in is disapper in this  case y -axis line is remove
# sns.despine(left=True)
plt.show()
```

```python
sns.set_context('notebook')
# style means specify the different marker like *,o
# alpha means dinsity of graph
sns.scatterplot(data=penguins,x='species',y='body_mass_g',style='sex',hue='island',alpha=0.5)
```

# Strip plot

```python
# dodge means seperate the same species different island with some distance
sns.stripplot(data=penguins,x='species',y='body_mass_g',hue='island',dodge=True,jitter=True)
```

# Swarm plot

```python
sns.swarmplot(data=penguins,x='species',y='body_mass_g',hue='island')
```

# Histogram

```python
# multiple means spearate based on hue
sns.histplot(data=penguins,x='body_mass_g',hue='sex',multiple='stack')
```

# Regression plot

```python
# scatter is false means the points  are not visible in graph
# sns.regplot(data=penguins,x='body_mass_g',y='flipper_length_mm',color='green',scatter=False)
sns.regplot(data=penguins,x='body_mass_g',y='flipper_length_mm',color='green',scatter=True)
```

# Line plot

```python
sns.set_style('whitegrid')
sns.lineplot(data=penguins,x='body_mass_g',y='flipper_length_mm',style='sex',hue='sex')
```

# Joint plot

```python
sns.set_style('white')
sns.jointplot(data=penguins,x='body_mass_g',y='flipper_length_mm')
```

```python
sns.jointplot(data=penguins,x='body_mass_g',y='flipper_length_mm',hue='sex')
```

```python
# kind means you can specify the plot are ['scatter', 'hist', 'hex', 'kde', 'reg', 'resid']
sns.jointplot(data=penguins,x='body_mass_g',y='flipper_length_mm',hue='island',kind='scatter')
```

```python
sns.jointplot(data=penguins,x='body_mass_g',y='flipper_length_mm',kind='reg')
```

# KED Plot

```python
sns.kdeplot(data=penguins,x='body_mass_g',hue='species',fill=True)
```

# Heat Map

```python
penguins.head()
```

```python
# taking all numeric column
column=['bill_length_mm',	'bill_depth_mm',	'flipper_length_mm',	'body_mass_g']
# if it possitive means increasing else decreasing
penguins[column].corr()
```

```python
# annot means that tell value in table
# cmap means change color of heatmap
# vmin means the negative corelation start with -0.2
# linewidth means space between box
# linecolor means color of the line
sns.heatmap(data=penguins[column].corr(),annot=True,cmap='Blues',vmin=-0.3,linewidths=2,linecolor='black')
```

# Rug plot

```python

sns.rugplot(data=penguins,x='body_mass_g',hue='species',palette='Set2',height=0.2)
# if need in y axis
# sns.rugplot(data=penguins,y='body_mass_g',hue='species',palette='Set2',height=0.2)
```

# Pair plot

```python
# diag_kind means diagonal in histogram
sns.pairplot(data=penguins,hue='sex',palette='Set2',diag_kind='hist')
```

# Pair Grid

```python
graph=sns.PairGrid(data=penguins,hue='sex',palette='Set2')
graph.map_upper(sns.scatterplot)
graph.map_lower(sns.kdeplot)
graph.map_diag(sns.histplot)
```
